```python
matrix = [[1, 2, 3],
          [4, 5, 6],
          [7, 8, 9]]

print(matrix[1][2])
```

```python
from operator import itemgetter

grades = [
    {'name': 'Joey', 'score': 80},
    {'name': 'King', 'score': 75},
    {'name': 'Larry', 'score': 90},
    {'name': 'John', 'score': 37},
    {'name': 'Ian', 'score': 60}
    ]

# key = lambda x:x['score']
grades.sort(key=itemgetter('score'), reverse=True)
print(grades)
```

```python
transposed = [
    [row[i] for row in matrix] for i in range(len(matrix[0]))
]

print(transposed)
```

```python
single_list = [element for row in matrix for element in row]
print(single_list)
```
