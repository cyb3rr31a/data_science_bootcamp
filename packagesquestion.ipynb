```python
current_date = dt.date.today()
print(f"Current date: {current_date}")

future_date = current_date + dt.timedelta(days=100)
print(f"100 days from today: {future_date}")

given_date = dt.date(2003, 7, 23)

day_of_week = given_date.strftime("%A")
print(f"The day of the week for {given_date} is {day_of_week}.")
```
