```python
age = (11, 13, 17, 16, 16, 13, 19, 15)
age_set = set(age)
print(age_set)
```

```python
def min_max_sum(tuple1: tuple):
    min = 1000
    max = 0
    sum = 0

    for i in tuple1:
        if i < min:
            min = i

        if i > max:
            max = i

        sum += i

    return min, max, sum

print(min_max_sum(age))
```
