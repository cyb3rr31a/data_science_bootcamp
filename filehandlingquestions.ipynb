```python
def read_file(filename):
    try:
        with open(filename, 'r') as file:
            for line in file:
                print(line, end='')
    except FileNotFoundError:
        print(f"The file {filename} does not exist.")

read_file("sample.txt")
```

```python
def write_file(string_list, filename):
        with open(filename, 'w') as file:
            for text in string_list:
                file.write(f"{text}\n")

list = ['hello', 'world', 'I', 'am', 'a', 'robot']

write_file(list, 'output.txt')
```

```python
def copy_file(filename1, filename2):
    with open(filename1, 'r') as file1, open(filename2, 'w') as file2:
        for line in file1:
            file2.write(line)

copy_file('source.txt', 'destination.txt')
```

```python
def append_to_file(text, filename):
    with open(filename, 'a') as f:
        f.write(text + '\n') 

append_to_file('world', 'log.txt')
```

```python
def count_words(filename):
    count = 0

    with open(filename, 'r') as f:
        for line in f:
            words = line.split()
            count += len(words)

    return count

count_words('document.txt')
```

```python
def find_and_replace(word1, word2, filename):
    with open(filename, 'r') as f:
        content = f.read()

        new_content = content.replace(word1, word2)

        with open(filename, 'w') as f:
            f.write(new_content)

find_and_replace('love', 'hate', 'data.txt') 
```

```python
def reverse_words(filename):
    try:
        with open(filename, 'r') as f:
            lines = f.readlines()

            for line in reversed(lines):
                print(line, end='')

    except FileNotFoundError:
        print(f"The file {filename} does not exist.")

reverse_words('reverse.txt')
```

```python
def lines_words_characters(filename):

    lines = 0
    words = 0
    characters = 0

    try:
        with open(filename, 'r') as f:
            for line in f:
                lines += 1
                words += len(line.split())
                characters += len(line)

        return lines, words, characters
                
    except FileNotFoundError:
        print(f"The file {filename} does not exist.")

stats = lines_words_characters('stats.txt')

if stats:
    print(f"Lines: {stats[0]}")
    print(f"Words: {stats[1]}")
    print(f"Characters: {stats[2]}")
```

```python
def merge_files(file_list, output_filename='merged.txt'):
    with open(output_filename, 'w') as outfile:
        for filename in file_list:
            with open(filename, 'r') as infile:
                outfile.write(infile.read())

                outfile.write('\n')

merge_files(['reverse.txt', 'sample.txt'])
```

```python
def split_file(filename='large.txt', chunk_size=100):
    try:
        with open(filename, 'r') as infile:
            lines = []
            file_count = 1

            for line in infile:
                lines.append(line)

                if len(lines) == chunk_size:
                    out_filename = f"part_{file_count}.txt"
                    with open(out_filename, 'w') as outfile:
                        outfile.writelines(lines)

                    lines = []
                    file_count += 1

            if lines:
                out_filename = f"part_{file_count}.txt"
                with open(out_filename, 'w') as outfile:
                    outfile.writelines(lines)

        print(f"Successfully split into {file_count} files.")

    except FileNotFoundError:
        print(f"The file {filename} does not exist.")

split_file('large.txt', 100)
```

```python
import datetime

def log_message(message, filename='activity.log'):
    timestamp = datetime.datetime.now().strftime("%Y-%m-%d %H:%M:%S")

    with open(filename, 'a') as log_file:
        log_file.write(f"[{timestamp}] {message}\n")

log_message("User logged in.")
log_message("Database backup started.")
```

```python
def copy_binary_file(source="image.bin", destination="copy_image.bin"):
    try:
        with open(source, 'rb') as infile, open(destination, 'wb') as outfile:
            chunk_size = 8192
            while True:
                chunk = infile.read(chunk_size)
                if not chunk:
                    break
                outfile.write(chunk)

    except FileNotFoundError:
        print(f"The file {source} does not exist.")

copy_binary_file('image.bin', 'copy_image.bin')
```

```python
import csv

def read_csv_to_dicts(filename="data.csv"):
    try:
        with open(filename, mode='r', newline='', encoding='utf-8') as file:
            reader = csv.DictReader(file)
            
            data = [row for row in reader]
        
            return data
            
    except FileNotFoundError:
        print(f"The file {filename} does not exist.")


read_csv_to_dicts('data.csv')
```

```python
import json

def read_json_to_dict(filename="data.json"):
    try:
        with open(filename, mode='r', encoding='utf-8') as file:
            data = json.load(file)
            return data
            
    except FileNotFoundError:
        print(f"The file {filename} does not exist.")
    except json.JSONDecodeError:
        print(f"The file {filename} is not valid JSON.")

read_json_to_dict('data.json')
```

```python
def read_protected_file(filename="protected.txt"):
    try:
        with open(filename, 'r') as file:
            content = file.read()
            return content

    except PermissionError:
        print(f"Permission denied: You do not have the required access rights to read '{filename}'.")

    except FileNotFoundError:
        print(f"The file '{filename}' does not exist.")

read_protected_file('protected.txt')
```
