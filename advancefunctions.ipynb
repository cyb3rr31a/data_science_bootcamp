```python
def fibonnaci(n):
    if n < 2:
        return n
    return fibonnaci(n - 1) + fibonnaci(n - 2)

fibonnaci(10)
```

```python
def add_to_dict(a: tuple, b=None):
    b = {}

    key, value = a
    b[key] = value

    return b

age = {}
add_to_dict(('Jenny', 17), age)
```

```python
def filter_integers(**kwargs):
    return {k: v for k, v in kwargs.items() if isinstance(v, int) and not isinstance(v, bool)}

result = filter_integers(
    age=30,           # Integer (Will be kept)
    is_active=True,   # Boolean (Will be filtered out)
    name="Alice",     # String (Will be filtered out)
    score=100,        # Integer (Will be kept)
    weight=65.5,      # Float (Will be filtered out)
    has_pet=False     # Boolean (Will be filtered out)
)

print(result)
```

```python
import time
from functools import wraps

def calculate_time(func):
    @wraps(func)
    def wrapper(*args, **kwargs):
        start_time = time.perf_counter()

        result = func(*args, **kwargs)

        end_time = time.perf_counter()
        execution_time = end_time - start_time

        print(f"[{func.__name__}] finished in {execution_time:.6f} seconds.")
        return result

    return wrapper

# Apply decorator to a complex calculation
@calculate_time
def count_primes(limit):
    count = 0
    for num in range(2, limit + 1):
        is_prime = True
        for i in range(2, int(num ** 0.5) + 1):
            if num % i == 0:
                is_prime = False
                break

        if is_prime:
            count += 1

    return count

print(f"Total primes: {count_primes(10_000)}\n")
print(f"Total primes: {count_primes(50_000)}\n")
print(f"Total primes: {count_primes(100_000)}\n")
```

```python
def filter_and_map(filter_func, map_func, numbers):
   return [map_func(num) for num in numbers if filter_func(num)]

raw_numbers = [-5, -2, 0, 1, 2, 3, 4, 5, 10, 15]

# Keep even numbers, then square them 
def is_even(n):
   return n % 2 == 0

def square(n):
   return n ** 2

result_1 = filter_and_map(is_even, square, raw_numbers)
print(f"Even squared: {result_1}")

# Keep positive numbers, then double them
result_2 = filter_and_map(lambda x: x > 0, lambda x: x * 2, raw_numbers)
print(f"Postives doubled: {result_2}")

# Keep numbers greater than 4, convert them to strings
result_3 = filter_and_map(lambda x: x > 4, lambda x: str(x), raw_numbers)
print(f"Greater than 4 as strings: {result_3}")
```

```python
def compose(f, g):
    def inner_function(x):
        return f(g(x))
    return inner_function

def get_first_odd(numbers):
    for num in numbers:
        if num % 2 != 0:
            return num
    return 0

def cube(n):
    return n ** 3

cube_first_odd = compose(cube, get_first_odd)

age = [11, 12, 14, 16]

print(f"First odd number cubed: {cube_first_odd(age)}")
```

```python
from functools import partial

def multiply(a, b):
    return a * b

double = partial(multiply, 2) # pre-fills the first argument 'a'

print(f"double(5) = {double(5)}")
print(f"double(-3) = {double(-3)}")
print(f"double(1.5) = {double(1.5)}")
```

```python
def average(numbers):
    try: 
        return sum(numbers) / len(numbers)
    except (ZeroDivisionError, TypeError):
        return None

print(average([10, 20, 30, 40]))
print(average([-5, 5, 10, 10]))
print(average([]))
print(average([10, 20, 'thirty']))
print(average([None]))
```

```python
def separate_types(items):
    integers = []
    strings = []
    floats = []

    for i in items:
        if isinstance(i, int) and not isinstance(i, bool):
            integers.append(i)
        elif isinstance(i, str):
            strings.append(i)
        elif isinstance(i, float):
            floats.append(i)

    return integers, strings, floats

raw_data = [10, "hello", 3.14, -5, "world", 0.99, True, 42, False]

ints, strs, flts = separate_types(raw_data)

print(f"Integers: {ints}")
print(f"Strings:  {strs}")
print(f"Floats:   {flts}")
```
