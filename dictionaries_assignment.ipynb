```python
squares = {
    1: 1,
    2: 4,
    3: 9,
    4: 16,
    5: 25,
    6: 36,
    7: 49,
    8: 64,
    9: 81,
    10: 100
}

print(squares)
```

```python
print(squares[5])
print(squares.keys())
```

```python
squares[11] = 121
del squares[1]

print(squares)
```

```python
for key, value in squares.items():
    print(f"{key}: {value}")
```

```python
cubes = {i: i*i*i for i in squares}
print(cubes)
```

```python
a = {
    1: 1,
    2: 4,
    3: 9,
    4: 16,
    5: 25 
}

b = {
    6: 36,
    7: 49,
    8: 64,
    9: 81,
    10: 100
}

print(a | b)
```

```python
ls_squares = []

for key, value in a.items():
    ls_squares.append((key, value))

print(ls_squares)
```

```python
even_nums = {key: value for key, value in squares.items() if key % 2 == 0}

print(even_nums)
```

```python
swapped = {value: key for key, value in a.items()}

print(swapped)
```

```python
from collections import defaultdict

hobbies = defaultdict(list)

hobbies['John'].append('Baking')
hobbies['Jane'].append('Biking')
hobbies['Jane'].append('Gaming')
hobbies['John'].append('Pottery')

print(dict(hobbies))
```

```python
def string_count(string):
    key = string
    value = len(string)


    return {key: value}

print(string_count('hello'))
```
